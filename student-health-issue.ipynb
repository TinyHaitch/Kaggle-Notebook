```python
# Two features
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(data['bmi'], data['heart_rate'],
           alpha=0.3, s=10, color='steelblue')
ax.set_xlabel('BMI')
ax.set_ylabel('Heart Rate')
ax.set_title('BMI vs Heart Rate - Isolated points are outliers')
plt.show()

# All feature pairs in a grid — scatter matrix
from pandas.plotting import scatter_matrix
scatter_matrix(data, figsize=(15, 15),
               alpha=0.2, diagonal='hist', s=5)
plt.suptitle('Scatter Matrix - All Feature Pairs')
plt.show()
```
